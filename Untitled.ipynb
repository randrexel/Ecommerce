```python
import pandas as pd
from sklearn.datasets import fetch_openml

print("Iniciando la descarga del dataset abierto KDDCup09 Churn (50k registros)...")

# Descarga directa desde OpenML bajo licencia abierta
kdd_churn = fetch_openml(name='KDDCup09_churn', version=1, as_frame=True, parser='auto')

# Convertir a DataFrame de Pandas
df = kdd_churn.frame

# Guardar el archivo CSV en tu directorio local
output_filename = 'Open_Subscription_Churn_50k.csv'
df.to_csv(output_filename, index=False)

print(f"¡Descarga completada con éxito!")
print(f"Archivo guardado como: '{output_filename}'")
print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas.")
```
